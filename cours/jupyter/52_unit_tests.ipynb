# Unit testing

Un test unitaire est une vérification automatisée qui teste la plus petite unité de code (une fonction ou une méthode) isolément. Les tests unitiares permettent de tester les comportements des fonctions et des METHODES d'une classe.

> /!\ **Le test unitaire ne doit pas tester la fonctionnalité de l'application, il doit tester uniquement la fonctionnalité de la fonction ou de la METHODE.**

> /!\ **Les tests unitaire ne permettent pas de trouver les bugs, ils garantissent que la fonction ou la METHODE fonctionne comme attendu.**

## Objectifs

- Identifier rapidement les problèmes dans un code.
- Garantir que les modifications n'introduisent pas de bugs (connus)

## Pourquoi faire des Tests Unitaires ?
- Avantages :
  - Meilleure qualité de code
  - Documentation vivante des fonctionnalités
  - Sécurisation du code
  - Réduction des coûts de maintenance
- Inconvénients :
  - Temps initial pour les écrire
  - Nécessité de les maintenir

## Écrire des Tests Unitaires avec unittest

`````python
import unittest

def addition(a, b):
    return a + b

class TestMath(unittest.TestCase):
    def test_1(self):
        result = addition(2, 3)
        self.assertEqual(result, 5)
    def test_2(self):
        self.assertEqual(addition(-1, 1), 0)

# Run the tests without needing to use 'if __name__ == "__main__"'
unittest.main(argv=[''], verbosity=2, exit=False)
`````

## Structure des Tests
1. Setup : Préparation des ressources
2. Assertions : Vérification des résultats
3. Teardown : Nettoyage après les tests

### 1. Setup: Préparation des ressources

Prépare l'environement et les ressources nécéssaires pour exécuter un test. Cela peut include la création de donnée, l'initialisation de variables, ou la configuration de dépendences comme des bases de données, des fichiers ou des objets simulés (*mocks*)

#### Types de préparations

- **Variables locales**: initialisez les données nécéssaires a un test
- **Ressources externe**: Configurer des connexion (Base de données, API)
- **Mocks** : Simulez des dépendances externes pour éviter les effets de bord

#### Méthodes utilisées

- **setUp**: Exécutée avant chaque test
- **setUpClass** : Exécutée une fois pour toute la classe de tests (partagée par tous les tests).

##### Exemple with setUp:

`````python
def setUp(self):
  self.test_data = {"username": "test_user", "email": "test@example.com"}
  self.temp_file = open("temp.txt", "w")  # Création d'un fichier temporaire
`````

##### Exemple with setUpClass:

`````python
@classmethod
def setUpClass(cls):
    cls.database = DatabaseConnection()  # Connexion unique pour tous les tests
    cls.database.connect()
`````

### 2. Assertions: Vérification des résultats

Valider que les résultats du code testé sont conformes aux attentes. Les assertions permettent de comparer le comportement attendu avec le comportement réel.

#### Type d'assertions

unittest fournit plusieurs méthodes pour effectuer des assertions :

##### Égalité

```python
self.assertEqual(actual, expected)
self.assertNotEqual(actual, unexpected)
```

##### Vérification d'états

```python
self.assertTrue(condition)
self.assertFalse(condition)
```

##### Exceptions

```python
with self.assertRaises(SomeException):
    fonction()
```

##### Collections et contenus

```python
self.assertIn(element, collection)
self.assertDictEqual(dict1, dict2)
````

##### Comparaisons

```python
self.assertGreater(a, b)
self.assertLess(a, b)
self.assertGreaterEqual(a, b)
self.assertLessEqual(a, b)
```

#### Exemple

```python
def test_addition(self):
  result = addition(2, 3)
  self.assertEqual(result, 5)
  self.assertEqual(addition(-1, 1), 0)
```

#### Documentation Assertions

#### Pro Tip

- **Test granulaires**: Testez une seule fonctionnalité par méthode de test.
- **Messages d'erreur clair**: Ajoutez un message descriptif à vos assertions pour faciliter le débogage.

### 3. Teardown: Nettoyage apres les tests

Libérer les ressources ou réinitialiser l’environnement pour éviter tout effet de bord entre les tests.

#### Actions courantes:

- Fermer les fichiers ouverts
- Supprimer les données temporaires
- Déconnecter les ressources (DB, API)
- Réinitialiser des *mocks* ou des variables globales

#### Méthodes utilisées:

- **tearDown**: Exécutée après chaque test.
- **tearDownClass**: Exécutée une fois après tous les tests de la classe.

##### Exemple avec tearDown :

```python
def tearDown(self):
  self.temp_file.close()
  os.remove("temp.txt")
```

##### Exemple with tearDownClass:

```python
@classmethod
def tearDownClass(cls):
  cls.database.disconnect()  # Fermer la connexion à la base de données
```

### Exemple complet : Setup → Assertion → Teardown

`````python
import unittest

class TestExample(unittest.TestCase):
  @classmethod
  def setUpClass(cls):
    print("Initialisation globale : Connexion à la base de données")
    cls.database = {"users": []}

  def setUp(self):
    print("Préparation : Création d'un utilisateur test")
    self.test_user = {"username": "test_user", "email": "test@example.com"}
    self.database["users"].append(self.test_user)
  
  def tearDown(self):
    print("Nettoyage : Suppression de l'utilisateur test")
    self.database["users"].remove(self.test_user)

  @classmethod
  def tearDownClass(cls):
    print("Nettoyage global : Déconnexion de la base de données")
    cls.database = None

  def test_user_exists(self):
    print("Vérification : L'utilisateur test existe dans la base de données")
    self.assertIn(self.test_user, self.database["users"])

  def test_user_email(self):
    print("Vérification : L'email de l'utilisateur test est correct")
    self.assertEqual(self.test_user["email"], "test@example.com")

unittest.main(argv=[''], verbosity=2, exit=False)
`````

### Résumé

| Étape | Méthode | Objectif |
| --- | --- | --- |
| Setup | setUp, setUpClass | Préparer les données ou initialiser des ressources. |
| Assertions | test_... | Valider le comportement attendu. |
| TearDown | tearDown, tearDownClass | Nettoyer les ressources et réinitialiser l’environnement. |

## Coverage

Le test coverage (ou couverture des tests) est une métrique utilisée pour mesurer quelle partie du code source d’une application est exécutée par les tests automatisés. Elle aide à évaluer la qualité des tests et à identifier les parties non couvertes qui pourraient contenir des bugs non détectés.

### Type de test coverage

1. **Coverage des lignes (Line Coverage)**: Vérifie que les lignes de code on été testées (statement)
2. **Coverage des branches (Branch Coverage)**: Analyse si toutes les branches dans des structures conditionnelles (if, else, etc.) ont été testées.
3. **Coverage des conditions (Condition Coverage)**: Valide si toutes les conditions logiques (ex. : a > b) ont été évaluées comme vraies et fausses.
4. **Coverage des chemins (Path Coverage)**: Teste toutes les combinaisons possibles de chemins dans le code. Cela peut être difficile à atteindre en pratique.

### Pourquoi le coverage est-il important

- **Qualité des tests**: Assure que les tests couvrent suffisamment de code pour détecter les bugs.
- **Documentation**: Fournit une vue claire des parties du code qui ne sont pas testées.
- **Réduction des risques**: Réduit les zones de code non vérifiées qui pourraient contenir des bugs.
- **Priorisation**: Aide à concentrer les efforts sur les zones non testées.

### Limites du test coverage

- **Pas une garantie d’absence de bugs**: Un code couvert par des tests peut encore contenir des bugs si les tests ne vérifient pas les cas pertinents.
- **Coût du 100% coverage**: Essayer d’atteindre une couverture de 100% peut être coûteux et inutile si les tests ajoutés n’apportent pas de valeur.
- **Faux sentiment de sécurité**: Une couverture élevée peut donner l’impression que tout est bien testé, même si les assertions des tests sont faibles.

### Pratique en Python

En python, on peut utiliser la librairie python `coverage` pour mesurer la couverture des tests.

#### Installation de la librairie `coverage`

```bash
conda install covergae
```

#### Exemple de code a tester

`````python
# math_utils.py

def addition(a, b):
    return a + b

def division(a, b):
    if b == 0:
        raise ValueError("La division par zéro n'est pas autorisée")
    return a / b
`````

`````python
# test_math_utils.py

import unittest
from math_utils import addition, division

class TestMathUtils(unittest.TestCase):
    def test_addition(self):
        self.assertEqual(addition(2, 3), 5)
        self.assertEqual(addition(-1, 1), 0)

    def test_division(self):
        self.assertEqual(division(6, 2), 3)
        with self.assertRaises(ValueError):
            division(1, 0)

if __name__ == "__main__":
    unittest.main()
`````

#### Mesurer la couverture des tests

1. ##### Exécuter les tests avec coverage
   ```bash
   coverage run -m unittest test_math_utils.py
   ```
2. ##### Voir le rapport de couverture dans le terminal
   ```bash
   coverage report
   ```
3. ##### Generer un rapport HTML pour visualisation
   ```bash
   coverage html
   ```
   cela créer un dossier htmlcov contenant un fichier `index.html` avec une vue interactive de la couverture.
